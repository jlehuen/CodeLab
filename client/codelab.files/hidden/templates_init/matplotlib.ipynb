## Tracé d'une sinusoïde avec Matplotlib

Cet exemple illustre le calcul et l'affichage graphique d'une fonction sinusoïdale $y = \sin(x)$ avec NumPy et Matplotlib.

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt

# Échantillonnage de 200 points sur l'intervalle [-2*pi, 2*pi]
x = np.linspace(-2 * np.pi, 2 * np.pi, 200)
y = np.sin(x)

# Tracé de la courbe sinusoïdale
plt.figure(figsize=(9, 4.5))
plt.plot(x, y, label="y = sin(x)", color="#1f77b4", linewidth=2)
plt.title("Fonction Sinusoïdale", fontsize=13, fontweight="bold")
plt.xlabel("x (radians)")
plt.ylabel("y")
plt.axhline(0, color="black", linewidth=0.8, linestyle="--", alpha=0.6)
plt.axvline(0, color="black", linewidth=0.8, linestyle="--", alpha=0.6)
plt.grid(True, linestyle=":", alpha=0.6)
plt.legend(loc="upper right")
plt.tight_layout()
plt.show()
```
