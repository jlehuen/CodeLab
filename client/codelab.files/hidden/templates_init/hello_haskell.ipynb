### Carnet interactif en langage Haskell (GHCi)
Chaque cellule de code Haskell peut définir des fonctions, des types ou évaluer directement des expressions et commandes `:t` / `:i`.
Lors de l'exécution d'une cellule, toutes les définitions des cellules précédentes sont automatiquement chargées !

```haskell
carre :: Int -> Int
carre x = x * x

:t carre
```

```haskell
putStrLn "Bonjour depuis le carnet Haskell de CodeLab !"
map carre [1..5]
```
